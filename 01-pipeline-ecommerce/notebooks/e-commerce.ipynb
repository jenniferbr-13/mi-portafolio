```python
#definir la función diagnosticar_tabla() que imprime un resumen rápido de calidad de datos de una tabla
def diagnosticar_tabla(nombre, df):
    """Imprime un resumen rápido de calidad de datos de una tabla."""
    print(f"\n{'='*50}")
    print(f"TABLA: {nombre}")
    print(f"{'='*50}")
    print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
    
    nulos = df.isnull().sum()
    nulos = nulos[nulos > 0]
    if len(nulos) > 0:
        print(f"\nNulos por columna:\n{nulos}")
    else:
        print("\nSin nulos")
    
    print(f"\nDuplicados: {df.duplicated().sum()}")
    print(f"\nTipos de datos:\n{df.dtypes}")
```
